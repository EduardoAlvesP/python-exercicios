    Uma lista de dicionários de vendas com loja, produto, valor e data (uns 10 r egistros), incluindo 2 inválidos de propósito: um com valor como texto ("abc") e outro sem a chave loja.

    Some o valor total por loja, ignorando os inválidos com try/except e contando quanto
    s foram ignorados.

    Encontre o produto com maior valor vendido em cada loja.

    Salve o resultado em resultado.json.

    Organize em funções, com uma main().

```python
import json
def somar_por_loja(lista_dicionarios):
  valor_por_loja = {}
  contador_ignorados = 0
  for dicionario in lista_dicionarios:
    try:
      valor_por_loja[dicionario['loja']] = valor_por_loja.get(dicionario['loja'],0) + dicionario['valor']
    except(KeyError,TypeError):
      contador_ignorados += 1
  return valor_por_loja,contador_ignorados





def produto_maior_valor(lista_dicionarios):
    valor_por_produto_loja = {}
    for dicionario in lista_dicionarios:
        try:
            loja = dicionario['loja']
            produto = dicionario['produto']
            valor = dicionario['valor']

            if loja not in valor_por_produto_loja:
                valor_por_produto_loja[loja] = {}

            produtos_da_loja = valor_por_produto_loja[loja]
            produtos_da_loja[produto] = produtos_da_loja.get(produto, 0) + valor
        except (TypeError, KeyError):
            pass

    resultado = {}
    for loja, produtos in valor_por_produto_loja.items():
        resultado[loja] = max(produtos, key=produtos.get)
    return resultado


def salvar_json(dados):
  with open("resultado.json", "w", encoding="utf-8") as f:
    json.dump(dados, f, ensure_ascii=False, indent=2)
```

```python

def main():
    lista_dicionarios = [
      {'loja':'rj','produto':'blush','valor':10,'data':'10:09:2026'},
      {'loja':'rj','produto':'A','valor':100,'data':'10:07:2026'},
      {'loja':'rj','produto':'A','valor':20,'data':'10:06:2026'},
      {'loja':'sp','produto':'B','valor':30,'data':'10:05:2026'},
      {'loja':'sp','produto':'D','valor':50,'data':'10:04:2026'},
      {'produto':'B','valor':60,'data':'10:03:2026'},
      {'loja':'sp','produto':'F','valor':'A','data':'10:02:2026'}]
    valor_loja, ignorados = somar_por_loja(lista_dicionarios)
    produtos_por_loja = produto_maior_valor(lista_dicionarios)
    resultado = {'total por loja':valor_loja,'registros ignorados':ignorados,'produto mais vendido por loja':produtos_por_loja}
    salvar_json(resultado)
```

```python
main()
```
